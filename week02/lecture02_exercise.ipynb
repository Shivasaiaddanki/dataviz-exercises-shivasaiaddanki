**Task 3 (stretch):** Build a **grouped bar chart** comparing 2 sub-factors (e.g. `GDP_per_capita` and `Freedom`) across the 5 most populated regions. Use colour meaningfully and write an insight title.

Regions to include: `'Western Europe'`, `'Latin America'`, `'East Asia'`, `'Sub-Saharan Africa'`, `'South Asia'`


```python

# Task 3: Grouped Bar Chart

# Step 1: Select five regions
selected_regions = [
    'Western Europe',
    'Latin America and Caribbean',
    'East Asia',
    'Sub-Saharan Africa',
    'South Asia'
]

# Step 2: Calculate regional averages
region_comparison = (
    df[df['Region'].isin(selected_regions)]
    .groupby('Region')[['GDP', 'Freedom']]
    .mean()
    .reindex(selected_regions)
    .reset_index()
)

# Step 3: Convert data into long format
region_long = region_comparison.melt(
    id_vars='Region',
    value_vars=['GDP', 'Freedom'],
    var_name='Factor',
    value_name='Score'
)

# Step 4: Create grouped bar chart
fig = px.bar(
    region_long,
    x='Region',
    y='Score',
    color='Factor',
    barmode='group',
    color_discrete_map={
        'GDP': '#2878B5',
        'Freedom': '#F39C35'
    },
    title='GDP and Freedom: Comparing Five World Regions',
    labels={
        'Region': 'World Region',
        'Score': 'Average Score',
        'Factor': 'Indicator'
    },
    text_auto='.2f'
)

# Step 5: Customize chart
fig.update_layout(
    template='plotly_white',
    height=600,
    bargroupgap=0.15,
    xaxis_tickangle=-20,
    legend_title='Indicator'
)

fig.show()

```

```python

# Build a horizontal bar chart
fig = px.bar(
    combined,
    x='Happiness_Score',
    y='Country',
    color='Group',
    orientation='h',
    color_discrete_map={
        'Top 8': '#2CA02C',
        'Bottom 8': '#D62728'
    },
    title='Top 8 vs Bottom 8 Happiest Countries',
    labels={
        'Happiness_Score': 'Happiness Score',
        'Country': 'Country',
        'Group': 'Country Group'
    },
    text='Happiness_Score'
)

# Add global average reference line
fig.add_vline(
    x=global_avg,
    line_dash='dash',
    line_color='black',
    annotation_text=f'Global Average: {global_avg:.2f}'
)

# Customize layout
fig.update_layout(
    template='plotly_white',
    height=650,
    xaxis_title='Happiness Score',
    yaxis_title='Country',
    legend_title='Group'
)

fig.update_traces(texttemplate='%{text:.2f}', textposition='outside')

fig.show()
```
